## 1. Problema, pregunta y decisión

### 1.1. Planteamiento del problema

Durante el Mundial 2026 circularon reportes en medios digitales sobre dificultades para seguir los partidos a través de ViX. Esa percepción motiva la revisión de los registros de afectaciones de clientes de **izzi Telecom**, el lugar donde trabajé desde septiembre del 2025 hasta septiembre del 2026, para conocer si el periodo del torneo presentó un comportamiento diferente al observado en los dos años anteriores.

En este proyecto analizo **fallas generales de izzi**, que pueden involucrar **Internet, Telefonía o Vídeo**. Los registros permiten estudiar las afectaciones asociadas a incidencias generales y su distribución temporal y territorial.

El problema consiste en contrastar la percepción de deterioro del servicio con evidencia sobre las fallas registradas. Para hacerlo, es necesario distinguir la existencia de afectaciones, su cantidad y su repetición frente al histórico. La pregunta no se resuelve únicamente contando registros del periodo del Mundial.

**La relación con ViX es el contexto que origina la investigación.** Una falla general de izzi no identifica por sí sola una interrupción de ViX ni demuestra que el cliente estuviera viendo un partido.


### 1.2. Pregunta principal

> **¿Se registraron afectaciones por fallas generales para clientes de izzi durante el Mundial 2026 y cómo se comparan su volumen y repetición con las mismas fechas de 2024 y 2025?**

La comparación permitirá examinar si los registros de izzi respaldan una percepción de aumento de afectaciones durante ese periodo. El alcance de la respuesta se refiere a las fallas generales registradas; valorar toda la experiencia de uso de ViX requiere información adicional sobre la plataforma y sus usuarios.

**Objetivo general:** comparar las afectaciones registradas por izzi durante el Mundial 2026 con las de 2024 y 2025, para evaluar el comportamiento histórico de las fallas de sus clientes.

**Objetivos específicos:**

- Comparar clientes afectados, incidencias generales, dispositivos identificados y combinaciones cliente–incidencia.
- Examinar la repetición de fallas por cliente afectado.
- Ubicar diferencias por entidad federativa, hub y plaza, además de la comparación nacional.
- Observar el comportamiento del periodo completo, de cada día y de las ventanas asociadas a los partidos.

### 1.3. Hipótesis de trabajo

El periodo del Mundial 2026 podría presentar más afectaciones o mayor repetición de fallas que las mismas fechas de 2024 y 2025.

También se contempla que los registros muestren un comportamiento similar o menor, o diferencias que dependan del territorio y del intervalo observado. Evalúo estas posibilidades en las secciones 7 y 11 y tomo mi decisión en las secciones 12 y 15. Respaldo el aumento descriptivo registrado y mantengo separada la causalidad, que mis fuentes no permiten identificar.


### 1.4. Decisión que se busca sustentar

La decisión es **determinar si el diagnóstico de un aumento de fallas generales para clientes de izzi durante el Mundial está respaldado por sus registros históricos**, antes de adoptar esa interpretación a partir de los reportes en medios digitales.

La lectura de los datos deberá permitir distinguir entre evidencia de aumento, un comportamiento similar o menor y evidencia insuficiente para responder en algún intervalo. Si aparecen diferencias, la interpretación debe conservar su ubicación temporal y territorial.

El juicio de que el servicio fue «tan malo como se decía» se concreta aquí en aspectos observables: **cantidad de clientes afectados, cantidad de incidencias y repetición de fallas**. Estos aspectos no equivalen a medir satisfacción, duración de las interrupciones o calidad de reproducción de ViX.

Un incremento frente a 2024 y 2025 tampoco demuestra que el Mundial haya causado las fallas. Del mismo modo, no observar un incremento agregado no descarta afectaciones particulares ni desmiente las experiencias individuales reportadas.


### 1.5. Referencia temporal del problema

La configuración del proyecto establece:

| Elemento | Referencia |
|---|---|
| Periodo objetivo | Del **11 de junio al 19 de julio de 2026**, ambos incluidos. |
| Referencias históricas | Del **11 de junio al 19 de julio de 2024** y de **2025**. |
| Comparación por partido | El mismo mes, día y hora local del inicio de cada partido, trasladados a 2024 y 2025. |
| Ventana de partido | Desde **30 minutos antes** hasta **180 minutos después** del inicio; el extremo final se excluye. |
| Referencia horaria | **America/Mexico_City**, como utiliza el programa. |

Se comparan las mismas fechas y horas locales; el programa no cambia las referencias para hacer coincidir el día de la semana. Las ventanas históricas son intervalos equivalentes de comparación, sin implicar que en 2024 y 2025 no se celebrara un mundial.

La comparación del periodo completo y las comparaciones por ventanas responden a preguntas distintas. Un cliente puede aparecer en varios días o ventanas, por lo que sus conteos no se suman para obtener el total de clientes distintos del periodo.

### 2.3. Qué representa una afectación en este proyecto

Una **incidencia general** se identifica mediante `WIDMT_INCIDENCIA_REMEDY`. Los registros de clientes permiten conocer qué identificadores de cliente están asociados a esa incidencia.

En mi análisis defino una **falla por cliente** como una combinación distinta de cliente e incidencia general. Una misma incidencia puede afectar a varios clientes y un mismo cliente puede estar asociado a varias incidencias.

En la operación de izzi, **solo un cliente activo puede presentar una falla general**. Por ello, interpreto la presencia de una afectación elegible como correspondiente a un cliente activo al momento de esa falla. Esta condición de negocio es distinta de la existencia de un registro en el CRM y no se deduce de `flg_eliminado = 0`.

| Servicio comprendido en los datos | Contexto para interpretar el problema |
|---|---|
| **Internet** | Es el servicio de conectividad. Una afectación de Internet y un error de una plataforma de streaming son eventos distintos que requieren identificación propia. |
| **Telefonía** | Corresponde al servicio de voz de izzi. Sus afectaciones forman parte del alcance general, sin demostrar una interrupción de la transmisión de un partido. |
| **Vídeo** | Corresponde al servicio de televisión/vídeo de izzi. Su inclusión en las fallas generales no permite equiparar cada registro con una falla de ViX. |

**En mi análisis agrego las fallas generales y no las desgloso por estos tres servicios.** Internet, Telefonía y Vídeo describen el alcance empresarial que considero; no los utilizo como una variable de clasificación en el programa ya que no pude conseguir esa data.

Esta distinción evita interpretar una afectación del conjunto como si necesariamente fuera una falla de Internet, de Vídeo o de ViX.

### 2.5. Variables derivadas para responder la pregunta

Las variables derivadas resumen los registros dentro de un intervalo y un territorio. Documento su significado y las utilizo después en mis tablas e interpretaciones.

| Variable | Definición | Qué permite observar |
|---|---|---|
| `CLIENTES_DISTINTOS` | Número de identificadores de cliente distintos afectados en el intervalo y territorio. | Alcance de las afectaciones entre los clientes registrados. |
| `INCIDENCIAS_DISTINTAS` | Número de incidencias generales distintas asociadas a esos clientes. | Cantidad de eventos generales registrados con afectaciones enlazadas. |
| `MAC_DISTINTAS` | Número de MAC distintas informadas. | Alcance entre los dispositivos identificados; no incluye MAC faltantes. |
| `FALLAS_CLIENTE` | Número de pares distintos `(WIDMT_CLIENTE, WIDMT_INCIDENCIA_REMEDY)`. | Afectaciones considerando que un cliente puede sufrir varias incidencias. |
| `FALLAS_PROMEDIO_POR_CLIENTE` | `FALLAS_CLIENTE / CLIENTES_DISTINTOS`, cuando el denominador es mayor que cero. | Repetición media de incidencias por **cliente afectado**; no es una tasa sobre toda la cartera activa. |
| `PROMEDIO_HISTORICO_<METRICA>` | Promedio de los conteos equivalentes de 2024 y 2025 para cada métrica base. | Referencia histórica conjunta, además de las comparaciones por año. |
| `DIF_<METRICA>_VS_<REFERENCIA>` | Conteo de 2026 menos el conteo de referencia. | Diferencia absoluta frente a 2024, 2025 o al promedio histórico. |
| `VAR_<METRICA>_PCT_VS_<REFERENCIA>` | `100 × (conteo_2026 − referencia) / referencia`. | Diferencia relativa; queda vacía cuando la referencia es cero o no está disponible. |
| `COBERTURA_COMPLETA` | Indicador de cobertura temporal utilizado por el programa para los intervalos comparados. | Distingue comparaciones con cobertura disponible; no garantiza que todos los eventos reales estén registrados. |

**Contexto del conteo:** un cliente asociado a dos incidencias aporta un cliente distinto y dos fallas cliente–incidencia. Si existen varias MAC para el mismo cliente y la misma incidencia, el par cliente–incidencia se cuenta una sola vez.

El programa también utiliza `CLIENTES_DISTINTOS`, `INCIDENCIAS_DISTINTAS` y `MAC_DISTINTAS` como variables de su evaluador de anomalías. Desarrollo su justificación, cálculo y sensibilidad en la sección 10.

Las variables utilizadas por el programa no identifican el consumo de ViX ni la contratación del pase, y no incorporan la duración de las interrupciones o el total de clientes activos. Por ello, estos conteos no constituyen una medida directa de disponibilidad de ViX ni de la probabilidad de falla de cualquier cliente de izzi. Una ausencia de fallas registradas tampoco demuestra por sí sola que ese cliente tuviera el servicio activo en el año correspondiente.


## 3. Población o universo

### 3.1. Universo organizacional y población observada

Para mi proyecto parto de un universo histórico de fallas de clientes de izzi que supera los **168 millones de registros**, con historia **desde febrero de 2021** y **con las reglas de negocio ya aplicadas**. Sobre ese universo apliqué el filtro temporal de este proyecto.

Utilizo las reglas de negocio y de asociación de clientes de los modelos productivos de **dbt para el Warehouse de izzi en Snowflake**. Los SQL de `anexos_sql/` son el resultado de su compilación. El código productivo parametriza las fechas para permitir el reprocesamiento; así aplico **los mismos filtros a la información de 2024, 2025 y 2026**. Documento este funcionamiento con mayor detalle en la sección 4.2.

Para definir la población de mi estudio distingo tres conjuntos:

| Conjunto | Delimitación |
|---|---|
| Universo histórico elegible del Warehouse | Más de 168 millones de registros de fallas de clientes, desde febrero de 2021, **después de las reglas de negocio y antes del filtro temporal que apliqué al proyecto**. No equivale a clientes ni a incidencias generales distintas. |
| Extracto temporal disponible | Las **64 partes** de afectaciones de clientes que conservan los años 2024, 2025 y 2026, del 10 de junio al 20 de julio, junto con las fuentes de incidencias generales y la dimensión territorial utilizadas para sus cruces. |
| Población efectivamente analizada | Registros de ese extracto que cumplen los requisitos del programa, se encuentran en los intervalos locales definidos y se enlazan con una incidencia general y el catálogo territorial. Los conteos se deduplican según la unidad correspondiente. |

Las reglas de negocio restringen las afectaciones de clientes a **Netcool y EAssure**, identificados mediante `submitter`, y a las asociaciones con clientes y equipos que cumplen los filtros y cruces del SQL anexo. Su detalle se presenta en la sección 4.2.

El universo observado queda condicionado por el registro organizacional, las reglas de asociación, la extracción y el filtro temporal. **En mi estudio describo las afectaciones identificadas en ese conjunto**, sin equipararlas a toda la cartera de clientes, a todos los eventos reales de la red o a los usuarios de ViX.

El filtro Netcool/EAssure se aplica a la construcción de las **afectaciones de clientes**. El SQL anexo de incidencias generales no contiene esa misma restricción de `submitter`; una incidencia solo participa en las métricas de afectación cuando tiene clientes elegibles enlazados. No documento aquí la fecha de corte del universo histórico ni su conteo exacto, por lo que no calculo una proporción de retención respecto de los más de 168 millones.


### 3.2. Unidades de observación y de análisis

Una fila física, un cliente y una falla general representan unidades distintas. Su identificación evita interpretar el volumen del extracto como cantidad de clientes o sumar conteos que se superponen.

| Unidad | Identificador o definición | Uso en el estudio |
|---|---|---|
| Registro de afectación | `WIDMT_INCIDENCIA_REMEDY_CLIENTE`. | Unidad de entrada de la fuente de clientes. El programa conserva una fila por identificador al cargarla; su cantidad no equivale al número de fallas cliente–incidencia. |
| Afectación cliente–incidencia | Par distinto `(WIDMT_CLIENTE, WIDMT_INCIDENCIA_REMEDY)`. | Unidad contada en `FALLAS_CLIENTE`. Varias filas o MAC del mismo par no crean nuevas afectaciones en esta métrica. |
| Cliente afectado | `WIDMT_CLIENTE` distinto dentro del intervalo y agrupación. | Unidad de `CLIENTES_DISTINTOS` y del análisis de repetición. El archivo por cliente reúne a quienes aparecen con afectaciones enlazadas en al menos uno de los tres periodos. |
| Incidencia general | `WIDMT_INCIDENCIA_REMEDY` distinto con afectaciones elegibles enlazadas. | Unidad de `INCIDENCIAS_DISTINTAS`. Una incidencia puede asociarse a muchos clientes. |
| Dispositivo identificado | `MAC` distinta informada en los registros seleccionados. | Unidad de `MAC_DISTINTAS`; una MAC no se interpreta como un cliente. Los registros sin MAC pueden contribuir a los otros conteos. |
| Intervalo–territorio | Combinación de periodo completo, día o ventana de partido, con nivel y etiqueta territorial; en la comparación se mantienen separados los años. | Unidad agregada para comparar los conteos. Las ventanas pueden compartir clientes e incidencias; estas filas no son observaciones independientes de personas o eventos. |

Por ejemplo, un cliente asociado a dos incidencias aporta **un cliente distinto y dos pares cliente–incidencia** en el intervalo que las contiene. Si una de esas incidencias aparece con dos MAC, el número de pares no aumenta por ese motivo.

Los clientes sin afectaciones registradas en ninguno de los periodos **no forman una población de control de clientes sin fallas**. Para construirla se necesitaría la cartera activa y su exposición al servicio en cada año. Un cero en un año del archivo por cliente significa ausencia de incidencias enlazadas registradas para ese identificador en ese periodo, con la cobertura disponible; no acredita que el cliente tuviera servicio activo.


### 3.3. Delimitación temporal y territorial

| Dimensión | Alcance de la población analizada |
|---|---|
| Periodo objetivo | **11 de junio a 19 de julio de 2026**, ambos incluidos, en `America/Mexico_City`. |
| Periodos históricos | Las mismas fechas de **2024 y 2025**, con sus horas locales equivalentes. No se sustituyen por otros periodos ni se ajustan al mismo día de la semana. |
| Intervalo técnico del periodo completo | Desde el **11 de junio a las 00:00:00** hasta el **20 de julio a las 00:00:00**, con el extremo final excluido, en cada año. El 20 de julio no añade un día al periodo analítico. |

| Territorios | Registros vinculados a `Dataset_Dim_Region.csv`, agrupados por **ENTIDAD_FEDERATIVA**, **HUB_CM** y **PLAZA**. El resultado nacional agrega el conjunto elegible, no acredita por sí solo cobertura de toda la red nacional. |

La configuración operacional se obtiene de `config.json`. El filtro previo de clientes conserva del **10 de junio al 20 de julio** en 2024, 2025 y 2026; ese margen se aplica sobre `FECHA_INCIDENCIA` antes de la conversión horaria realizada por el programa. Los límites definitivos se seleccionan después en hora local. Los días del margen no se usan como controles adicionales.

La población del periodo completo, la de un día y la de una ventana pueden superponerse. Cada conteo de clientes, incidencias, MAC o pares se obtiene dentro de su propio intervalo y agrupación. **No se suman los clientes distintos de días, territorios o partidos para obtener el total nacional del periodo.**


### 3.4. Inclusiones y exclusiones

El alcance empresarial comprende fallas generales relacionadas con **Internet, Telefonía y Vídeo**, como se documentó en la sección 2.3. El programa las agrega; no aplica una selección analítica de usuarios de ViX ni presenta una clasificación por esos servicios.

| Criterio | Efecto sobre la población |
|---|---|
| Remitentes de las afectaciones | Selecciono Netcool y EAssure por ser fuentes de calidad. Excluyo otros remitentes para reducir el riesgo de incidencias de usuarios no verídicas o duplicadas respecto de las que envían esas fuentes. |
| Identificación y asociación en el Warehouse | Se requieren las correspondencias entre incidencia, identificador de equipo, dimensión de clientes y cuenta organizacional que establece el SQL. Una falla sin esas correspondencias no llega a su salida. |
| Filtros de clientes y equipos | Quedan fuera los valores de cliente expresamente excluidos, equipos con `orig_id` distinto de 3 o con `flg_eliminado` distinto de 0, y asociaciones que no satisfacen las condiciones de vigencia descritas en 4.2. |
| Fechas del análisis | Quedan fuera de cada comparación los registros cuya fecha local no pertenece a su intervalo. El extracto con margen no amplía automáticamente la población analítica. |
| Campos requeridos | El cargador omite registros sin los campos obligatorios de afectación, incidencia, cliente, región o fecha. **MAC es opcional**; su ausencia no elimina el registro de los conteos de clientes o incidencias. |
| Incidencia general enlazada | Las afectaciones sin correspondencia con una incidencia general se excluyen de las métricas y se contabilizan en el control de calidad. Las incidencias generales sin afectaciones elegibles tampoco contribuyen a esas métricas. |
| Catálogo territorial | Una llave de región sin correspondencia provoca la detención del programa. Una etiqueta territorial vacía de una llave válida se conserva como **NO IDENTIFICADA**. |
| Cartera activa y uso de ViX | El estudio no incorpora como población a todos los clientes activos ni identifica espectadores, contratantes o usuarios de ViX. |

En mis datos, **`orig_id = 3` identifica el CRM de izzi**; no es una categoría de tecnología. El SQL incluye `tipo_tecnologia` en su salida, pero no utilizo esa variable para desglosar el análisis.

Documento `flg_eliminado = 0` como una condición de existencia del registro en el CRM, **no como un indicador de cliente activo**. En el SQL anexo, el filtro se aplica a `ttc_dim_equipos`, dentro de `dim_equipos_filtro`.

Distingo la existencia del registro en el CRM de la actividad del cliente: **solo un cliente activo puede presentar una falla general**. Los clientes activos sin afectaciones registradas no aparecen por ese motivo en la población analizada, y los registros del CRM no sustituyen a un padrón de cartera activa.


### 3.5. Qué población permiten describir los resultados

## 6. Preparación y limpieza

### 6.1. Reglas de lectura y campos requeridos

La limpieza tiene etapas con efectos diferentes: elegibilidad organizacional en dbt, selección temporal del extracto publicado y preparación de los CSV filtrados en el programa. Las reglas de dbt ya se desarrollaron en 4.2; aquí consolido las reglas que determinan la utilización analítica del extracto.

| Regla vigente | Tratamiento implementado | Justificación y alcance |
|---|---|---|
| Lectura del CSV | Codificación `utf-8-sig`; normalización de nombres de columnas con `strip().upper()`; reconocimiento de `,`, `|`, `;` o tabulador. La dimensión se lee con `|`. | Reconoce las columnas sin depender de mayúsculas, espacios del encabezado o BOM; el separador se adapta al archivo. |
| Estructura | Falta de columnas requeridas, encabezados repetidos o archivo vacío producen error. | Evita continuar con una estructura que no permite identificar o relacionar los registros. |
| Campos de clientes | Son obligatorios afectación, incidencia, cliente, región y fecha. Una fila con alguno vacío se omite y suma a `filas_incompletas_omitidas`. | Son necesarios para ubicar la fila y calcular sus unidades. La omisión se registra como una exclusión de datos. |
| MAC de clientes | La columna debe existir, pero el valor puede faltar. El valor vacío se almacena como `NULL`; no se imputa ni se elimina por ese motivo al cliente. | Permite conservar clientes, incidencias y pares aunque no se identifique el dispositivo. `COUNT(DISTINCT mac)` excluye los `NULL`. |
| Incidencias generales | Identificador vacío: se omite la fila. La fecha debe corresponder al año del archivo y comenzar con el formato esperado `AAAA-MM-DD`. | Permite un enlace con identificador y una partición temporal coherente. Estas comprobaciones no son una validación exhaustiva de todas las fechas de origen. |
| Dimensión | Llaves de región vacías o repetidas producen error; etiquetas vacías se sustituyen por `NO IDENTIFICADA`. | Exige una correspondencia unívoca de llave y conserva una categoría explícita cuando falta la etiqueta. |
| Calendario | Identificador vacío o repetido, fechas u horas no convertibles, partidos fuera del periodo o cantidad distinta de 104 producen error. | Evita ventanas repetidas o incompletas respecto del calendario configurado. No reemplaza la validación con su fuente. |

El cargador normaliza **los encabezados**, pero no aplica una normalización general de los valores de identificadores o MAC. No elimina separadores de una MAC, no convierte sistemáticamente sus valores a mayúsculas ni interpreta automáticamente textos como `NULL`, `NA` o `N/A` como faltantes. En 6.5 documento el perfil observado para esos casos; no atribuyo al código una limpieza que no realiza.


### 6.2. Fechas, zonas horarias e intervalos

`FECHA_INCIDENCIA` procede del timestamp que el programa interpreta en **UTC**. `fecha_irc_local` utiliza `datetime.fromisoformat`; si el valor no contiene zona, le asigna UTC, y si contiene un desplazamiento, lo convierte primero a UTC. Después transforma el instante a `America/Mexico_City` y lo guarda en hora local con precisión de segundos.

`FECHA_ENVIO` ya está en hora de México conforme al SQL de incidencias generales. El programa no vuelve a aplicarle la conversión UTC de clientes. Esta diferencia evita comparar una fecha UTC con una hora local como si fueran la misma referencia.

| Etapa temporal | Tratamiento | Motivo |
|---|---|---|
| Filtro previo | 10 de junio a 20 de julio de los tres años, sobre la columna de cada fuente; las fechas ilegibles no cumplen la máscara. | Reduce el extracto y conserva margen para la selección local posterior. |
| Prefiltro de clientes en el cargador | Comparación conservadora del prefijo de fecha UTC, ampliado dos días alrededor de los límites de carga. | Evita convertir fechas que están lejos de los intervalos. El margen técnico no se incorpora al periodo analítico. |
| Selección definitiva de clientes | Después de convertir a México, aplica `inicio <= fecha_mx < fin`. | Ubica la afectación dentro de su intervalo local sin incluir dos veces un extremo compartido. |
| Selección de incidencias generales | Prefiltro por fecha local, con las fechas de ambos extremos incluidas. | Conserva incidencias para los cruces; puede retener el día del extremo final sin ampliar las afectaciones contadas. |
| Periodo y días | Periodo del 11 de junio a las 00:00 hasta el 20 de julio a las 00:00, fin excluido; cada día se construye con la misma convención. | El periodo analítico comprende del 11 de junio al 19 de julio. |
| Partidos | Inicio del partido menos 30 minutos; final más 180 minutos, excluido. Misma fecha y hora local en 2024 y 2025. | Conserva la comparación definida, sin sustituirla por otra fecha o día de la semana. |

Con el calendario y la configuración revisados, `limites_carga` produce exactamente los intervalos **11 de junio 00:00 a 20 de julio 00:00, fin excluido**, de cada año. En una configuración distinta, la envolvente de carga podría extenderse para incluir ventanas; los intervalos finales seguirían seleccionándose por separado.

Una fecha de clientes que falla en la conversión después de pasar el prefiltro provoca un error con el archivo correspondiente. Las filas que no pasan el prefiltro se contabilizan como fuera de los límites; no se afirma que se haya validado su fecha completa. El filtro previo declarado convierte las fechas ilegibles a `NaT` y las excluye; distingo ese tratamiento del error que produce el cargador definitivo.

La conciliación horaria del control de calidad compara hasta **500 incidencias** enlazadas; el reporte disponible registró **0 diferencias superiores a un minuto** en las 500. El código detiene la ejecución si más del 1 % de las comparaciones difiere en más de 60 segundos. No es una muestra aleatoria ni una auditoría de todas las fechas: para cada incidencia se toma una afectación sin un criterio explícito de prioridad temporal.


### 6.3. Deduplicación y atributos contradictorios

El programa distingue deduplicar filas al cargar de contar unidades distintas al analizar:

| Unidad o tabla | Regla actual | Si dos filas con la misma llave tienen atributos diferentes |
|---|---|---|
| Afectación de cliente, tabla `afectados` | Llave primaria `WIDMT_INCIDENCIA_REMEDY_CLIENTE`; `INSERT OR IGNORE`. | Se conserva **la primera fila aceptada**. Las siguientes se ignoran aunque cambien cliente, incidencia, región, MAC o fecha local. No se combinan atributos ni se elige automáticamente la fecha más reciente. |
| Incidencia general, tabla `incidencias` | Llave primaria `WIDMT_INCIDENCIA_REMEDY`; `INSERT OR REPLACE`. | Se conserva **la última fila aceptada** para ese identificador, incluida su `FECHA_ENVIO`. Es la última en orden de lectura, no necesariamente la de fecha cronológicamente mayor. |
| Dimensión y calendario | Llaves vacías o repetidas se rechazan. | El programa detiene la carga; no selecciona silenciosamente una versión. |
| Clientes, incidencias y dispositivos | `COUNT(DISTINCT ...)` dentro de cada intervalo y agrupación. | Se cuentan los valores conservados tras la carga; esta agregación no resuelve un conflicto perdido en la deduplicación anterior. |

Las partes de clientes se leen en orden numérico del 1 al 64; las particiones generales, por año. Dentro de cada archivo se conserva su orden de filas. Las reglas de primera y última fila proporcionan un comportamiento reproducible para el mismo contenido y orden, pero **no acreditan qué atributo sería correcto si hubiera un conflicto**. El índice final no conserva las versiones ignoradas o reemplazadas. En una ejecución que sí presente reducción de filas, ese índice por sí solo no permitiría cuantificar conflictos: necesitaría conservar o revisar las filas competidoras del extracto filtrado antes de deduplicar y definir el criterio de prioridad que utilizaré. ### 6.5. Perfil de MAC y etiquetas territoriales

Trato MAC como un identificador opcional: una afectación sin ese dato puede contribuir a clientes, incidencias y pares, mientras que `COUNT(DISTINCT mac)` excluye los valores almacenados como `NULL`. La carga normaliza encabezados, pero no corrige sistemáticamente formatos o tokens de los valores.

En las tablas agregadas cuantifico MAC distintas y su relación con clientes en la sección 7. No uso la diferencia entre ambos conteos para estimar registros sin MAC: un cliente puede tener varios dispositivos y los denominadores son distintos. La validez operacional de una MAC, la correspondencia única con equipos y el perfil de las filas excluidas requieren comprobaciones adicionales.

Si necesito perfilar las filas cargadas, puedo consultar el índice derivado mediante una conexión SQLite de sólo lectura con `mode=ro` y `PRAGMA query_only=ON`:

```sql
SELECT substr(fecha_mx, 1, 4) AS anio,
       COUNT(*) AS registros,
       SUM(mac IS NULL) AS mac_null,
       SUM(mac IS NOT NULL AND trim(mac) = '') AS mac_vacia_espacios,
       SUM(mac IS NOT NULL AND mac != trim(mac)) AS mac_espacios_borde,
       SUM(mac IS NOT NULL AND upper(trim(mac))
           IN ('NULL', 'NONE', 'NAN', 'NA', 'N/A')) AS mac_tokens
FROM afectados
GROUP BY substr(fecha_mx, 1, 4);
```

Esta consulta describe una comprobación adicional y no forma parte de las celdas ejecutadas del documento. No presento sus resultados como mediciones realizadas por esta copia.

Para el territorio, una llave inexistente detiene el análisis; una etiqueta vacía de una llave válida se conserva como `NO IDENTIFICADA`. Distingo esa categoría semántica de la falta de correspondencia de la llave y no imputo una ubicación. Mis resultados se refieren a las agrupaciones conservadas en el catálogo utilizado.

### 10.2. Cómo reproduzco el cálculo

Para las dos referencias $(r_{2024},r_{2025})$, calculo por componente $(j)$:

$$
c_j=\operatorname{mediana}(r_{2024,j},r_{2025,j}),\qquad
MAD_j=\operatorname{mediana}(|r_{a,j}-c_j|),\qquad
s_j=\max(1,1{,}4826\,MAD_j,\sqrt{c_j+1}).
$$

Transformo cada vector como $(z_j=(x_j-c_j)/s_j)$. Calculo la distancia
euclídea de 2026 a cada referencia y promedio sus $(k)$ distancias menores.
Para cada referencia calculo también la distancia a sus vecinos excluyéndose
a sí misma. Con esas distancias base $(D_a)$ obtengo:

$$
S=\frac{D_{2026}-\operatorname{mediana}(D_a)}
{\max(1,1{,}4826\operatorname{mediana}(|D_a-\operatorname{mediana}(D_a)|))}.
$$

Clasifico como `ANOMALIA` sólo si se cumplen simultáneamente: clientes de 2026
≥ 10; exceso sobre la mediana histórica ≥ 10; clientes de 2026 estrictamente
mayores que el P95 histórico de clientes; distancia actual estrictamente mayor
que el P95 de las distancias base; y **S ≥ 3**. El P95 de clientes se redondea
a dos decimales antes de la comparación, de acuerdo con el código. No interpreto
ese percentil con dos observaciones como una prueba de significancia.

Con dos referencias y **k=1**, las dos distancias base son iguales; su MAD vale
cero y el denominador de S queda en uno. Si ambos históricos tienen cero clientes,
las tres variables históricas son cero, la escala es uno y una observación
actual con al menos diez clientes satisface las reglas vigentes. Por eso una
puntuación muy alta puede resultar de la escala y de referencias cero;
**S no es una probabilidad ni un número de desviaciones estándar**.
